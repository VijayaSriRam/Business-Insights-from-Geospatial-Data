```python
# Importing the required modules

from unqlite import UnQLite
from math import cos, sin, sqrt, atan2, radians

# Access the data collection after connecting to UnQLite DB

db = UnQLite('sample.db')
data = db.collection('data')
```

```python
# Defining distance function algorithm. It is used to calculatethe distance between two locations

def DistanceFunction(latitude2, longitude2, latitude1, longitude1):
    R = 3959
    pi1 = radians(latitude1)
    pi2 = radians(latitude2)
    deltaPi = radians(latitude2-latitude1)
    deltaLambda = radians(longitude2-longitude1)
    a = (sin(deltaPi/2) * sin(deltaPi/2)) + (cos(pi1) * cos(pi2) * sin(deltaLambda/2) * sin(deltaLambda/2))
    c = 2 * atan2(sqrt(a), sqrt(1-a))
    d = R * c
    return d
```

```python
# Defining FindBusinessBasedOnCity function which is used to identify all the businesses present in the given city

def FindBusinessBasedOnCity(cityToSearch, saveLocation1, collection):    
    businesses = []
    for idx in range(len(collection.all())):
        dataAtIndex = collection.fetch(idx)      
        if dataAtIndex['city'] == cityToSearch:
            businesses.append([dataAtIndex['name'], dataAtIndex['full_address'], dataAtIndex['city'], dataAtIndex['state']]) 
    f = open(saveLocation1, 'w')
    for attr in businesses:
        f.write('$'.join(str(s) for s in attr))
        f.write('\n')
    f.close()

# Defining FindBusinessBasedOnLocation function which is used to identify all the businesses present within a certain range from the specified location


def FindBusinessBasedOnLocation(categoriesToSearch, myLocation, maxDistance, saveLocation2, collection):
    names = set([])

    lat1 = myLocation[0]
    lon1 = myLocation[1]
    
    for idx in range(len(collection.all())):
        data_point = collection.fetch(idx)
        lat2 = data_point['latitude']
        lon2 = data_point['longitude']

        # filter by location and category
        distance = DistanceFunction(lat2, lon2, lat1, lon1)
        if distance <= maxDistance:
            for category in categoriesToSearch:
                if category in data_point['categories']:
                    names.add(data_point['name'])

# Save the above search queries to a file

    f = open(saveLocation2, 'w')
    for name in names:
        f.write(name + '\n')
    f.close()
```
